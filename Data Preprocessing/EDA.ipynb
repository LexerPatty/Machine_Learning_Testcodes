# Univariate Analysis: Analysis of ONE variable.
# Bivariate Analysis: Analysis of TWO variables.
# Multivariate Analysis: Analysis of MORE THAN TWO variables.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Create Dataset
data = {
    "Name": [
        "Rahul", "Amit", "Priya", "Raj", "Sneha",
        "Rohan", "Anjali", "Karan", "Pooja", "Arjun"
    ],

    "Study_Hours": [2, 3, 5, 7, 6, 4, 8, 3, 6, 5],

    "Attendance": [60, 70, 85, 90, 95, 75, 98, 65, 88, 80],

    "Sleep_Hours": [5, 6, 7, 8, 7, 6, 8, 5, 7, 6],

    "Marks": [40, 50, 75, 85, 90, 60, 95, 45, 88, 70],

    "Age": [22, 23, 21, 24, 22, 23, 21, 25, 22, 24],

    "Gender": [
        "Male", "Male", "Female", "Male", "Female",
        "Male", "Female", "Male", "Female", "Male"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Column	Data Type	Univariate Analysis
# Name	Categorical (Identifier)	Unique values / Frequency
# Study_Hours	Numerical	Mean, Median, Mode, Histogram, Boxplot
# Attendance	Numerical	Mean, Median, Mode, Histogram, Boxplot
# Sleep_Hours	Numerical	Mean, Median, Mode, Histogram, Boxplot
# Marks	Numerical	Mean, Median, Mode, Histogram, Boxplot
# Age	Numerical	Mean, Median, Mode, Histogram, Boxplot
# Gender	Categorical	Mode, Countplot
```

```python
# Name Analysis

print("Total Students:", df["Name"].count())

print("Unique Names:", df["Name"].nunique())

print("Names:")
print(df["Name"].unique())

print("Name Frequency:")
print(df["Name"].value_counts())
```

```python
# Analyze the numerical column 

numerical_cols = df.select_dtypes(include=np.number).columns
print(numerical_cols)
```

```python
for col in numerical_cols:
    print(f"\nAnalysis of {col}")

    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:", df[col].mode()[0])
```

```python
for col in numerical_cols:
    
    plt.figure(figsize=(7, 4))

    sns.histplot(data=df, x=col, bins=5, kde=True)

    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")

    plt.show()
```

```python
for col in numerical_cols:
    
    plt.figure(figsize=(7, 4))

    sns.boxplot(data=df, x=col)

    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)

    plt.show()
```

```python
# Gender Frequency
print(df["Gender"].value_counts())

# Gender Mode
print("Gender Mode:", df["Gender"].mode()[0])

# Countplot
plt.figure(figsize=(7, 4))

sns.countplot(data=df, x="Gender")

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

Bivariate Analysis

# Bivariate Analysis means analyzing two variables (columns) together to understand the relationship between them. 

# In simple words, we check whether changes in one column are related to changes in another column.
# Real-life example
# Imagine you want to understand:
# - Does studying more hours increase students' marks?
# - Does attendance affect students' marks?
# - Is there any relationship between sleep hours and marks?

```python
df
```

```python
# Example 1: Study Hours vs Marks
# Question: Do students who study more hours get higher marks?


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours",
    y="Marks",
    s=100
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
# B. Correlation
# Correlation measures how strongly two numerical variables are related.

correlation = df["Study_Hours"].corr(df["Marks"])
print("Correlation:", correlation)
```

```python
# Example 2: Attendance vs Marks
# Question: Do students with higher attendance get higher marks?

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance",
    y="Marks",
    s=100
)

plt.title("Attendance vs Marks")
plt.xlabel("Attendance (%)")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
print(df["Attendance"].corr(df["Marks"]))
```

```python
# Example 3: Sleep Hours vs Marks
# Question: Is there a relationship between sleep hours and marks?

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Sleep_Hours",
    y="Marks",
    s=100
)

plt.title("Sleep Hours vs Marks")
plt.xlabel("Sleep Hours")
plt.ylabel("Marks")

plt.grid(True)
plt.show()
```

```python
print(df["Sleep_Hours"].corr(df["Marks"]))
```

```python
# 4. Categorical vs Numerical Analysis
# Here, we compare one categorical column with one numerical column.

# Example 4: Gender vs Marks
# Question: How do marks vary between male and female students?


plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Marks"
)

plt.title("Gender vs Marks")
plt.xlabel("Gender")
plt.ylabel("Marks")

plt.show()
```

```python
# Example 5: Gender vs Study Hours

plt.figure(figsize=(8, 5))

sns.barplot(
    data=df,
    x="Gender",
    y="Study_Hours",
    errorbar=None
)

plt.title("Average Study Hours by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Study Hours")

plt.show()
```

```python
df["Result"] = [
    "Fail", "Fail", "Pass", "Pass", "Pass",
    "Pass", "Pass", "Fail", "Pass", "Pass"
]

df
```

```python
# Example 6: Gender vs Result
# Question: How many male and female students passed or failed?

result_table = pd.crosstab(
    df["Gender"],
    df["Result"]
)

print(result_table)
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    result_table,
    annot=True,
    cmap="Blues",
    fmt="d"
)

plt.title("Gender vs Result")
plt.show()
```

# What is Multivariate Analysis?

# Multivariate Analysis means analyzing three or more variables (columns) together to understand their combined relationships and patterns.


# In simple words, instead of checking only two columns, we analyze multiple columns together to understand how they interact.
# Real-life example
# Imagine you want to understand what affects a student's marks.

# You might consider:
# - Study Hours
# - Attendance
# - Sleep Hours
# - Age
# - Marks
# Instead of comparing only Study Hours vs Marks, we analyze all these variables together.

# This is called Multivariate Analysis.

```python
# Correlation Heatmap
# Question: How are Study Hours, Attendance, Sleep Hours, Age, and Marks related to each other?
# A correlation heatmap helps us understand the relationships between multiple numerical columns at once.

# Select numerical columns automatically
numerical_cols = df.select_dtypes(include=np.number)

# Calculate correlation
correlation_matrix = numerical_cols.corr()

# Plot heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap - Student Performance")
plt.show()
```

```python
# How to understand it?

# Correlation	Meaning
# Close to +1	Strong positive relationship
# Close to 0	Weak or no linear relationship
# Close to -1	Strong negative relationship

# For example:
# - Study Hours vs Marks: A positive correlation suggests that students studying more tend to have higher marks.
# - Attendance vs Marks: A positive correlation suggests higher attendance tends to be associated with higher marks.
# - Sleep Hours vs Marks: Shows whether sleep duration is associated with marks.
```
