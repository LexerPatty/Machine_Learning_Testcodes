# An outlier is a data point that is very different from the rest of the data.

# Example :
# Health Monitoring → If your heart rate usually is 70–80 bpm but suddenly jumps to 200, that’s an outlier → medical alert

```python
import matplotlib.pyplot as plt

# Normal heart rate readings (in bpm)
heart_rates = [72, 75, 78, 74, 76, 73, 77, 180, 195, 200]  # 200 is unusual

# Detect outliers (greater than 120 bpm)
outliers = [x for x in heart_rates if x > 120]

print("Heart Rate Readings:", heart_rates)
print("Detected Outliers (Medical Alert):", outliers)
```

```python
# Plot
plt.plot(heart_rates, marker="o", linestyle="-", color="blue", label="Heart Rate")
for idx, val in enumerate(heart_rates):
    if val in outliers:
        plt.scatter(idx, val, color="red", s=150, label="Outlier (Medical Alert)")

plt.xlabel("Reading Number")
plt.ylabel("Heart Rate (bpm)")
plt.legend()
plt.show()
```
