# For Regression :

# - Mean Absolute Error (MAE)
# - Mean Squared Error (MSE)
# - Root Mean Squared Error (RMSE)
# - R² Score

```python
# Mean Absolute Error (MAE)

# Simple Example:
# Suppose we want to predict house prices.

# Actual Prices (y_true): [200, 220, 250, 270]
# Predicted Prices (y_pred): [210, 215, 240, 280]

# Step 1 → Find absolute errors:
# |200 - 210| = 10
# |220 - 215| = 5
# |250 - 240| = 10
# |270 - 280| = 10

# Step 2 → Take average:

from sklearn.metrics import mean_absolute_error

# Actual and predicted values
y_true = [200, 220, 250, 270]
y_pred = [210, 215, 240, 280]

# Calculate MAE
mae = mean_absolute_error(y_true, y_pred)
print("Mean Absolute Error:", mae)
```

```python
# Mean Squared Error (MSE)

# Simple Example:
# Suppose we are predicting house prices:

# Actual Prices (y_true): [200, 220, 250, 270]
# Predicted Prices (y_pred): [210, 215, 240, 280]

# Step 1 → Find the errors:
# 200 - 210 = -10
# 220 - 215 = 5
# 250 - 240 = 10
# 270 - 280 = -10

# Step 2 → Square the errors:
# (-10)² = 100
# (5)² = 25
# (10)² = 100
# (-10)² = 100

# Step 3 → Take average:

from sklearn.metrics import mean_squared_error

# Actual and predicted values
y_true = [200, 220, 250, 270]
y_pred = [210, 215, 240, 280]

# Calculate MSE
mse = mean_squared_error(y_true, y_pred)
print("Mean Squared Error:", mse)
```

```python
# Root Mean Squared Error (RMSE)
# RMSE (Root Mean Squared Error) = √MSE

import numpy as np

rmse = np.sqrt(mse)
print(round(rmse, 2))
```

# R² Score

# It tells us how well a model’s predictions match the actual data.

# Interpretation :
# R² = 1 → Perfect prediction (model explains 100% of the variance).
# R² = 0 → Model is no better than just predicting the average.
# R² < 0 → Model is worse than the average.

# Suppose you’re predicting house prices from size.
# If R² = 0.9 → Your model explains 90% of the variation in prices.
# If R² = 0.2 → Only 20% of variation is explained; poor model.

# need :
# When building a regression model, we want to know how well the model explains the variation in the data.
# RMSE, MAE, MSE tell us about errors, but R² tells us how much of the data’s variance the model captures.
# It answers: “Does my model actually explain the trend, or is it just guessing around the mean?”

```python
import numpy as np
from sklearn.metrics import r2_score

# Actual values (y_true)
y_true = [3, 5, 7, 9]

# Predicted values (y_pred)
y_pred = [2.8, 5.1, 6.9, 9.2]

# Worst case :
actual_worst = [1000, 1200, 1500, 1800]
predicted_worst = [3000, 2800, 2700, 2600]

# R² Score
r2 = r2_score(y_true, y_pred)
print("R² Score:", r2)
print("In Percentage R² Score:", round(r2 *100 , 2),"%")

# R² Score worst case :
r21 = r2_score(actual_worst, predicted_worst)
print("\nR² Score worst case :", r21)
```
