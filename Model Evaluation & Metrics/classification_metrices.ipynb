```python
# Confusion Matrix

# |                 | Predicted Positive  | Predicted Negative  |
# | ----------------| ------------------- | ------------------- |
# | Actual Positive | True Positive (TP)  | False Negative (FN) |
# | Actual Negative | False Positive (FP) | True Negative (TN)  |
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt

# Actual labels (1 = Spam, 0 = Not Spam)
y_actual = [1, 0, 1, 1, 0, 0, 1, 0, 1, 0]

# Predicted labels
y_pred = [1, 1, 1, 0, 0, 0, 1, 0, 0, 0]

# Confusion Matrix
CM = confusion_matrix(y_actual, y_pred, labels=[1, 0])

disp = ConfusionMatrixDisplay(confusion_matrix=CM, display_labels=["Spam", "Not Spam"])

# classification_report
report = classification_report(y_actual, y_pred, target_names=["Not Spam", "Spam"])

disp.plot()
plt.show()

print("Confusion Matrix:\n", CM)
print("\nClassification Report:\n", report)
```

# Behind the scene

# | #  | Actual | Predicted |     Outcome         |
# | -- | :----: | :-------: | ------------------: |
# | 1  |    1   |     1     |  TP (true positive) |
# | 2  |    0   |     1     | FP (false positive) |
# | 3  |    1   |     1     |          TP         |
# | 4  |    1   |     0     | FN (false negative) |
# | 5  |    0   |     0     |  TN (true negative) |
# | 6  |    0   |     0     |          TN         |
# | 7  |    1   |     1     |          TP         |
# | 8  |    0   |     0     |          TN         |
# | 9  |    1   |     0     |          FN         |
# | 10 |    0   |     0     |          TN         |

# Totals:
# TP = 3 (rows 1,3,7)
# FP = 1 (row 2)
# FN = 2 (rows 4,9)
# TN = 4 (rows 5,6,8,10)

```python
# Extract values from confusion matrix
TP = CM[0,0]
FN = CM[0,1]
FP = CM[1,0]
TN = CM[1,1]

# Accuracy :
accuracy = (TP + TN) / (TP + TN + FP + FN)

# precision :
precision = TP / (TP + FP) if (TP+FP) != 0 else 0

# recall :
recall = TP / (TP + FN) if (TP+FN) != 0 else 0

# f1_score :
f1_score = 2 * (precision * recall) / (precision + recall) if (precision+recall) != 0 else 0

# error rate :
error_rate = 1 - accuracy

print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1 Score: {f1_score:.2f}")
print(f"Error Rate: {error_rate:.2f}")
```
