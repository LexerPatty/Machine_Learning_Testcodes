```python
import numpy as np

# Marks of two students
student_A = [6, 7, 8, 9, 11, 10 , 12]

student_B = [2,3,4,5,8,9,32]

# Mean
mean = np.mean(student_A)
print("Mean:", mean)
```

```python
# Population Variance
var = np.var(student_A)
print("Variance:",var )
```

```python
# Population Standard Deviation
std= np.std(student_A)
print("Standard Deviation:",std )
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm


# Calculate mean and standard deviation
mean = np.mean(student_A)
std = np.std(student_A)

# Generate X values
x = np.linspace(0, 100, 1000)

# Calculate normal distribution curve
curve_A = norm.pdf(x, mean, std)

# Plot
plt.figure(figsize=(12, 6))

plt.plot(
    x, curve_A,
    label=f"Student A (SD = {std:.2f})",
    linewidth=2
)

# Plot actual data points on the curve
y_values = norm.pdf(student_A, mean, std)

plt.scatter(
    student_A, y_values,
    color="red",
    s=60,
    label="Actual Marks"
)

# Display values above each point
for mark, probability in zip(student_A, y_values):
    plt.annotate(
        f"{mark}",
        (mark, probability),
        textcoords="offset points",
        xytext=(0, 10),
        ha="center"
    )

# Mean line
plt.axvline(
    mean,
    linestyle="--",
    label=f"Mean = {mean:.2f}"
)

# Standard deviation lines
plt.axvline(mean - std, linestyle=":", label=f"Mean - SD = {mean-std:.2f}")
plt.axvline(mean + std, linestyle=":", label=f"Mean + SD = {mean+std:.2f}")

# Titles and labels
plt.title("Variance and Standard Deviation - Bell Curve")
plt.xlabel("Marks")
plt.ylabel("Probability Density")

plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
