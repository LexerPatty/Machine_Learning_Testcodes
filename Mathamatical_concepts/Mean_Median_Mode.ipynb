```python
# pip install numpy pandas scikit-learn

import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
```

```python
data = {
    "Marks": [10, 20, 20, 30, 40, 50, 70, None]
}

df = pd.DataFrame(data)

df
```

```python
# Mean Imputation
mean_imputer = SimpleImputer(strategy="mean")

mean_result = mean_imputer.fit_transform(df)

print("Mean Imputed:", mean_result.ravel())
```

```python
mean = np.mean(df["Marks"])
print("Mean:", mean)
```

```python
# Median Imputation
median_imputer = SimpleImputer(strategy="median")

median_result = median_imputer.fit_transform(df)

print("Median Imputed:", median_result.ravel())
```

```python
median = np.median(df["Marks"])

median_imr= np.median(median_result)

print("Median:", median)
print(median_imr)
```

```python
mode_imputer = SimpleImputer(strategy="most_frequent")

mode_result = mode_imputer.fit_transform(df)

print("Mode Imputed:", mode_result.ravel())
```

```python
mode = df["Marks"].mode()
print("Mode:", mode.tolist())
```
