```python

import itertools

# =====================================
# STEP 1: GENERATE SAMPLE SPACE
# =====================================

coin = ['H', 'T']

sample_space = list(
    itertools.product(coin, repeat=3)
)

print("SAMPLE SPACE:")
print(sample_space)

print("Total Outcomes:", len(sample_space))
```

```python
# =====================================
# STEP 2: EVENT E (AT LEAST 2 TAILS)
# =====================================

E = []

for outcome in sample_space:

    tails_count = outcome.count('T')

    if tails_count >= 2:

        E.append(outcome)

print("\nEVENT E:")
print(E)

print("Total E Outcomes:", len(E))
```

```python
# =====================================
# STEP 3: EVENT F (FIRST TOSS IS HEAD)
# =====================================

F = []

for outcome in sample_space:

    if outcome[0] == 'H':

        F.append(outcome)

print("\nEVENT F:")
print(F)

print("Total F Outcomes:", len(F))
```

```python
# =====================================
# STEP 4: INTERSECTION (E AND F)
# =====================================

intersection = []

for outcome in E:

    if outcome in F:

        intersection.append(outcome)

print("\nINTERSECTION (E AND F):")
print(intersection)

print("Total Common Outcomes:", len(intersection))
```

```python
# =====================================
# STEP 5: CALCULATE PROBABILITIES
# =====================================

total = len(sample_space)

P_E = len(E) / total

P_F = len(F) / total

P_E_intersection_F = len(intersection) / total

print("\nINDIVIDUAL PROBABILITIES:")

print("P(E):", P_E)

print("P(F):", P_F)

print("P(E ∩ F):", P_E_intersection_F)
```

```python
# =====================================
# STEP 6: CONDITIONAL PROBABILITY
# =====================================

# Formula: P(E|F) = P(E ∩ F) / P(F)

if P_F != 0:

    P_E_given_F = P_E_intersection_F / P_F

else:

    P_E_given_F = 0

print("\nCONDITIONAL PROBABILITY:")

print("P(E|F) =", P_E_given_F)

print("Percentage:", P_E_given_F * 100, "%")
```

```python
# =====================================
# STEP 7: FINAL RESULT
# =====================================

print("\nFINAL RESULT:")

print(
    "Probability of at least 2 tails,",
    "given the first toss is Head =",
    P_E_given_F
)
```
