# Logistic Regression

# Logistic Regression is used when the output is Yes/No, 0/1, True/False (i.e., classification, not prediction of continuous numbers).
# Instead of predicting marks like linear regression, it predicts the probability of something happening.
# It uses the sigmoid function to squeeze predictions into the range 0 to 1.

# Real-life Example :
# Imagine we want to predict if a student passes (1) or fails (0) an exam based on hours studied.
# If probability ≥ 0.5 → Pass (1)
# If probability < 0.5 → Fail (0)

image.png

```python
import warnings
warnings.filterwarnings("ignore", category=UserWarning)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression

# Sample dataset
data = {
    'Hours_Studied': [2,3,4,5,6,7,8],
    'Pass_Fail':     [0,0,0,1,1,1,1]  # 0 = Fail, 1 = Pass
}

# Create DataFrame
df = pd.DataFrame(data)
```

```python
# Features (X) and Target (y)
X = df[['Hours_Studied']]
y = df['Pass_Fail']


# Train Logistic Regression Model
model = LogisticRegression()
model.fit(X, y)


# Predict for a new student
new_student = [[6.5]]   # hours studied
probability = model.predict_proba(new_student)   # gives probability
prediction = model.predict(new_student)          # gives class (0/1)


print("Probability of Passing:", probability)
print("Probability of Passing: {:.2f}%".format(probability[0][1] * 100))    # Access the probability of the 'Pass' class (index 1)
print("Probability of Fail: {:.2f}%".format(probability[0][0] * 100))     # Access the probability of the 'fail' class (index 0)
print("Predicted Class (0=Fail, 1=Pass):", prediction)

# probability[0][0] → chance of Fail
# probability[0][1] → chance of Pass
```
