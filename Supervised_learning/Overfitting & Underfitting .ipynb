```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

# -----------------------
# Example Data: Temperature vs Ice Cream Sales
temperature = np.array([5, 10, 15, 20, 25, 30, 35, 40]).reshape(-1, 1)
sales = np.array([5, 15, 40, 70, 90, 100, 85, 50])  # Sales rise then fall
```

```python
# -----------------------
# Models
underfit_model = make_pipeline(PolynomialFeatures(1), LinearRegression())  # Linear
goodfit_model = make_pipeline(PolynomialFeatures(2), LinearRegression())  # Quadratic
overfit_model = make_pipeline(PolynomialFeatures(7), LinearRegression())  # Too complex

underfit_model.fit(temperature, sales)
goodfit_model.fit(temperature, sales)
overfit_model.fit(temperature, sales)
```

```python
# -----------------------
# Predictions
x_test = np.linspace(0, 45, 200).reshape(-1, 1)
y_underfit = underfit_model.predict(x_test)
y_goodfit = goodfit_model.predict(x_test)
y_overfit = overfit_model.predict(x_test)
```

```python
# -----------------------
# Plot
plt.figure(figsize=(10,6))
plt.scatter(temperature, sales, color="black", label="Actual Sales Data")

plt.plot(x_test, y_underfit, "r-", label="Underfit (Linear)")
plt.plot(x_test, y_goodfit, "g-", label="Good Fit (Quadratic)")
plt.plot(x_test, y_overfit, "b-", label="Overfit (High Degree)")

plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales")
plt.title("Underfitting vs Overfitting: Ice Cream Sales Example 🍦")
plt.legend()
plt.show()
```
