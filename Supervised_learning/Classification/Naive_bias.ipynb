# Naive Bayes

# It’s based on Bayes’ Theorem, which calculates the probability of an event given some evidence its a classification algorithm .
# The “naive” part means it assumes all features are independent of each other — even if they are not.

image.png

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Sample data
emails = ["Win money now", "Hello friend", "Free prize", "How are you"]
labels = ["Spam", "Not Spam", "Spam", "Not Spam"]

# Convert text to numbers
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(emails)

# Train Naive Bayes
model = MultinomialNB()
model.fit(X, labels)

# Predict new email
new_email = ["Loose a free money"]
X_new = vectorizer.transform(new_email)
prediction = model.predict(X_new)

print("Predicted label:", prediction[0])
```

# Example: Colored Marbles

# Scenario:
# You have a bag of marbles from two bags: Bag A and Bag B.
# Each bag has red, blue, and green marbles.

# | Bag | Red | Blue | Green | Total |
# | --- | --- | ---- | ----- | ----- |
# | A   | 3   | 2    | 1     | 6     |
# | B   | 1   | 3    | 2     | 6     |

# Problem:

# You randomly pick a marble and it’s red. Which bag is it more likely to have come from?

# Step 1: Prior Probability
# P(Bag A) = 0.5 (since 2 bags, equal chance)
# P(Bag B) = 0.5

# Step 2: Likelihood (Naive Bayes)
# P(Red | Bag A) = 3/6 = 0.5
# P(Red | Bag B) = 1/6 ≈ 0.167


image.png

```python
from sklearn.naive_bayes import CategoricalNB

# Simple dataset
X = [[0],[0],[0],[1],[1],[2],[2],[1],[2],[0]]  # 0=Red, 1=Blue, 2=Green
y = ['A','A','A','A','A','A','B','B','B','B']  # Bag labels

# Train Naive Bayes
model = CategoricalNB()
model.fit(X, y)

# Predict the bag of a new marble
new_marble = [[0]]  # Red marble
prediction = model.predict(new_marble)
probabilities = model.predict_proba(new_marble)

print("Predicted Bag:", prediction[0])

prob_a = float(probabilities[0][0])*100
prob_b = float(probabilities[0][1])*100
print(f"Probability of Bag A: {prob_a:.2f}")
print(f"Probability of Bag B: {prob_b:.2f}")
```
