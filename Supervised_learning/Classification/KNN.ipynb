# k-Nearest Neighbors (KNN)

# What is KNN?
# KNN is a supervised machine learning algorithm used for classification (like Yes/No, categories) and regression (predict numbers).
# It predicts the label or value of a new data point by looking at its “K nearest neighbors” in the dataset.
# “Nearest” is calculated using distance, usually Euclidean distance.

# How it Works (Simple Steps) :
# Choose K (number of neighbors).
# Measure the distance between the new point and all points in the dataset.
# Select the K nearest points.
# Take a majority vote (for classification) or average (for regression) among neighbors.
# Assign the predicted label/value.

# Simple Analogy :
# Imagine you move to a new neighborhood and want to guess the favorite ice cream flavor of a new kid.
# You check the 3 nearest kids around them.
# If 2 of 3 like chocolate and 1 likes vanilla → you predict the new kid likes chocolate.

```python
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier

# Simple dataset
data = {
    'Movie': ['Jawan','Gadar 2','Rocky and rani ','OMG 2'],
    'IMDB_Rating': [8.0, 6.2, 7.2, 8.2],
    'Duration': [160, 170, 168, 155],
    'Genre': ['Action','Action','Comedy', 'Comedy']
}


df = pd.DataFrame(data)
```

```python
# Features and target
X = df[['IMDB_Rating', 'Duration']]
y = df['Genre']


# Train KNN
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X, y)



# Predict genre of a new movie
new_movie = [[7.4, 114]]   # IMDB Rating = 7.4, Duration = 114 mins
predicted_genre = knn.predict(new_movie)



print("Predicted Genre for the new movie Barbie is :", predicted_genre[0])
```
