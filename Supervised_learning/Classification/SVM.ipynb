# SVM Support vector machine

# SVM is a supervised machine learning algorithm used mainly for classification (and sometimes regression).
# It tries to find a decision boundary (hyperplane) that best separates different classes.
# The best boundary is the one that maximizes the margin between the classes.
# It works well for high-dimensional data and non-linear separation using kernels.

# Key Concepts :
# Hyperplane: The line (in 2D) or plane (in higher dimensions) that separates classes.
# Support Vectors: The data points closest to the hyperplane. They determine the position of the hyperplane.
# Margin: Distance between the hyperplane and the nearest points of each class. SVM maximizes this margin.

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import SVC
import pandas as pd

# -------------------------
# Sample dataset
data = {
    'Text': ['photo', 'picture', 'selfie',   # Image words
             'movie', 'clip', 'film',        # Video words
             'song', 'melody', 'tune'],      # Music words
    'Category': ['Image','Image','Image',
                 'Video','Video','Video',
                 'Music','Music','Music']
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# -------------------------
# Convert text to numerical features
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['Text']).toarray()
y = df['Category']


# -------------------------
# Train SVM classifier
model = SVC(kernel='linear', probability=True)
model.fit(X, y)


# -------------------------
# Predict new samples
new_samples = ['portrait', 'track', 'video']
X_new = vectorizer.transform(new_samples).toarray()
predictions = model.predict(X_new)
probabilities = model.predict_proba(X_new)


# -------------------------
# Print predictions and probabilities in %
for text, pred, prob in zip(new_samples, predictions, probabilities):
    prob_dict = {cls: round(p*100, 1) for cls, p in zip(model.classes_, prob)}
    print(f"Text: {text} -> Predicted: {pred}, Probabilities: {prob_dict}")
```
